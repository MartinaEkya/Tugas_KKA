```python
# latihan_3.py

class Hero:
    def __init__(self, name, hp, attack_power):
        self.name = name
        self.hp = hp
        self.attack_power = attack_power

    def info(self):
        print(f"Hero: {self.name} | HP: {self.hp} | Power: {self.attack_power}")

    def serang(self, lawan):
        print(f"{self.name} menyerang {lawan.name}!")
        lawan.diserang(self.attack_power)

    def diserang(self, damage):
        self.hp -= damage
        print(f"{self.name} terkena damage {damage}. Sisa HP: {self.hp}")


class Mage(Hero):
    def __init__(self, name, hp, attack_power, mana):

        self.mana = mana

    def info(self):
        print(f"{self.name} [Mage] | HP: {self.hp} | Mana: {self.mana}")

    def skill_fireball(self, lawan):
        if self.mana >= 20:
            print(f"{self.name} menggunakan Fireball ke {lawan.name}!")
            self.mana -= 20
            lawan.diserang(self.attack_power * 2)
        else:
            print(f"{self.name} gagal skill! Mana tidak cukup.")


if __name__ == "__main__":
    eudora = Mage("Eudora", 80, 30, 100)
    balmond = Hero("Balmond", 200, 10)

    eudora.info()
    eudora.serang(balmond)
    eudora.skill_fireball(balmond)
```

2. Kenapa bisa error padahal sudah memasukkan "Eudora"?

Karena "Eudora" memang dikirim saat membuat objek, tetapi data tersebut seharusnya diproses oleh constructor dari class Hero. Saat super().__init__() dihapus, constructor Hero tidak dijalankan. Akibatnya, name, hp, dan attack_power tidak dibuat di objek Mage.

3. Apa fungsi super()?

super() digunakan untuk memanggil constructor dari class induk, yaitu Hero. Dengan adanya super(), data seperti nama, HP, dan attack power dari Hero bisa digunakan oleh Mage. Jadi, super() membantu class Mage menggunakan bagian dari constructor milik Hero.

```python
# LATIHAN 4 - Enkapsulasi (Mengamankan Data HP)

class Hero:
    def __init__(self, nama, hp_awal):
        self.nama = nama
        # Enkapsulasi: HP bersifat Private (hanya bisa diakses di dalam class ini)
        self.__hp = hp_awal

    # GETTER: Cara resmi melihat HP
    def get_hp(self):
        return self.__hp

    # SETTER: Cara resmi mengubah HP (dengan validasi)
    def set_hp(self, nilai_baru):
        if nilai_baru < 0:
            self.__hp = 0   # HP tidak boleh negatif
        elif nilai_baru > 1000:
            print("Cheat terdeteksi! HP dimaksimalkan ke 1000 saja.")
            self.__hp = 1000
        else:
            self.__hp = nilai_baru

    def diserang(self, damage):
        # Kita pakai setter/getter bahkan di dalam class sendiri agar aman
        sisa_hp = self.get_hp() - damage
        self.set_hp(sisa_hp)
        print(f"{self.nama} terkena damage {damage}. Sisa HP: {self.get_hp()}")


if __name__ == "__main__":
    hero1 = Hero("Layla", 100)

    # hero1.__hp = 9999   # GAGAL (Tidak akan mengubah hp asli)
    # print(hero1.__hp)   # ERROR (Tidak bisa dibaca langsung)

    hero1.set_hp(-50)          # Coba set negatif
    print(hero1.get_hp())      # Output: 0 (Karena dicegat oleh logika Setter)

    # ---- Tugas Analisis 4.1: Percobaan Hacking (Name Mangling) ----
    print(f"Mencoba akses paksa: {hero1._Hero__hp}")
```

1. Apakah nilai HP muncul atau error? Mengapa?

Nilai HP tetap muncul. Hal ini karena Python mengubah nama __hp menjadi _Hero__hp, sehingga sebenarnya HP masih bisa diakses dengan cara tersebut. Tapi cara ini sebaiknya tidak digunakan karena kita jadi mengakses data private secara langsung.

2. Apa yang terjadi jika validasi pada set_hp() dihapus?

```python
# LATIHAN 5 - Abstraction & Interface (Membuat Kontrak/Standar)

from abc import ABC, abstractmethod

class GameUnit(ABC):

    @abstractmethod
    def serang(self, target):
        pass

    @abstractmethod
    def info(self):
        pass


class Hero(GameUnit):
    def __init__(self, nama):
        self.nama = nama

    def serang(self, target):
        print(f"Hero {self.nama} menebas {target}!")

    def info(self):
        print(f"Saya adalah Hero: {self.nama}")


class Monster(GameUnit):
    def __init__(self, jenis):
        self.jenis = jenis

    def serang(self, target):
        print(f"Monster {self.jenis} menggigit {target}!")

    def info(self):
        print(f"Saya adalah Monster: {self.jenis}")


if __name__ == "__main__":

    h = Hero("Alucard")
    m = Monster("Serigala")

    h.info()
    m.info()
    h.serang("Serigala")
    m.serang("Alucard")
```

1. Apa yang terjadi jika method serang() pada Hero dihapus?

Program akan error karena Hero belum memenuhi aturan yang sudah dibuat oleh GameUnit. Method serang() wajib ada, jadi kalau dihapus, objek Hero tidak bisa dibuat.

2. Mengapa GameUnit tidak bisa dibuat menjadi objek? Apa gunanya?

Karena GameUnit adalah abstract class, jadi memang hanya digunakan sebagai kerangka atau aturan dasar, bukan untuk dibuat menjadi objek langsung. Gunanya supaya class seperti Hero dan Monster memiliki method yang wajib dibuat, yaitu serang() dan info().

```python
# LATIHAN 6 - Polymorphism (Fleksibilitas Interaksi)
class Hero:
    def __init__(self, nama):
        self.nama = nama

    def serang(self):
        print("Hero menyerang dengan tangan kosong.")


class Mage(Hero):
    def serang(self):
        print(f"{self.nama} (Mage) menembakkan Bola Api! Boom!")


class Archer(Hero):
    def serang(self):
        print(f"{self.nama} (Archer) memanah dari jauh! Jleb!")


class Fighter(Hero):
    def serang(self):
        print(f"{self.nama} (Fighter) memukul dengan pedang! Slash!")


class Healer(Hero):
    def serang(self):
        print(f"{self.nama} tidak menyerang, tapi menyembuhkan teman!")


if __name__ == "__main__":
    pasukan = [
        Mage("Eudora"),
        Archer("Miya"),
        Fighter("Zilong"),
        Mage("Gord"),
        Healer("Estes"),
    ]

    print("--- PERANG DIMULAI ---")
    # Satu perintah loop, tapi respon berbeda-beda (Polymorphism)
    for pahlawan in pasukan:
        pahlawan.serang()
```

1. Apakah program tetap berjalan jika ditambahkan Healer? Apa keuntungannya?

Iya, program tetap berjalan selama Healer memiliki method serang(). Kita juga nggak perlu mengubah bagian looping-nya.

Menurut saya, keuntungan polymorphism adalah kita jadi lebih mudah menambahkan karakter baru. Misalnya nanti ingin menambahkan Healer, kita cukup membuat class-nya dan memasukkannya ke dalam list tanpa harus mengubah kode looping yang sudah ada.

2. Apa yang terjadi jika method serang pada Archer diubah menjadi tembak_panah?

Program akan error karena di bagian looping masih memanggil pahlawan.serang(), sedangkan Archer sudah tidak punya method serang().

Jadi, nama method harus tetap sama, yaitu serang(), supaya semua karakter bisa dipanggil dengan perintah yang sama meskipun cara menyerangnya berbeda.

```python

#Tugas Kelompok
#Intan Arum (12) dan Martina Ekya (17)
from abc import ABC, abstractmethod


# =========================================================
# UTILITAS: Format angka jadi format Rupiah (Rp 1.500.000)
# =========================================================
def format_rupiah(angka):
    return "Rp " + f"{angka:,.0f}".replace(",", ".")


# =========================================================
# ABSTRACTION: Kerangka dasar semua jenis kamar
# =========================================================
class KamarHotel(ABC):
    def __init__(self, nama_kamar, tipe, harga_dasar):
        self.nama_kamar = nama_kamar
        self.tipe = tipe                    # "Deluxe" / "Standard" (untuk pesan stok)
        self.__harga_dasar = harga_dasar    # ENCAPSULATION: private
        self.__stok = 0                     # ENCAPSULATION: private

    # ---------- GETTER ----------
    def get_stok(self):
        return self.__stok

    def get_harga_dasar(self):
        return self.__harga_dasar

    # ---------- SETTER (dengan validasi) ----------
    def tambah_stok(self, jumlah):
        if jumlah < 0:
            print(f"Gagal update stok Kamar {self.tipe}! Stok tidak boleh negatif ({jumlah}).")
            return False
        self.__stok += jumlah
        print(f"Berhasil menambahkan stok kamar {self.tipe}: {jumlah} unit.")
        return True

    # ---------- ABSTRACT METHOD (wajib diisi anak class) ----------
    @abstractmethod
    def tampilkan_detail(self, nomor, jumlah_malam):
        pass

    @abstractmethod
    def hitung_harga_total(self, jumlah_malam):
        pass


# =========================================================
# INHERITANCE + POLYMORPHISM: Kamar Deluxe
# =========================================================
class KamarDeluxe(KamarHotel):
    PAJAK = 0.10  # 10%

    def __init__(self, nama_kamar, harga_dasar, fasilitas):
        super().__init__(nama_kamar, "Deluxe", harga_dasar)
        self.fasilitas = fasilitas

    def hitung_harga_total(self, jumlah_malam):
        harga = self.get_harga_dasar()
        pajak = harga * self.PAJAK
        return (harga + pajak) * jumlah_malam

    def tampilkan_detail(self, nomor, jumlah_malam):
        harga = self.get_harga_dasar()
        pajak = harga * self.PAJAK
        subtotal = self.hitung_harga_total(jumlah_malam)
        print(f"{nomor}. [DELUXE] {self.nama_kamar} | Fasilitas: {self.fasilitas}")
        print(f"   Harga Dasar/Malam: {format_rupiah(harga)} | Pajak(10%): {format_rupiah(pajak)}")
        print(f"   Menginap: {jumlah_malam} malam | Subtotal: {format_rupiah(subtotal)}")
        return subtotal


# =========================================================
# INHERITANCE + POLYMORPHISM: Kamar Standard
# =========================================================
class KamarStandard(KamarHotel):
    PAJAK = 0.05  # 5%

    def __init__(self, nama_kamar, harga_dasar, kapasitas):
        super().__init__(nama_kamar, "Standard", harga_dasar)
        self.kapasitas = kapasitas

    def hitung_harga_total(self, jumlah_malam):
        harga = self.get_harga_dasar()
        pajak = harga * self.PAJAK
        return (harga + pajak) * jumlah_malam

    def tampilkan_detail(self, nomor, jumlah_malam):
        harga = self.get_harga_dasar()
        pajak = harga * self.PAJAK
        subtotal = self.hitung_harga_total(jumlah_malam)
        print(f"{nomor}. [STANDARD] {self.nama_kamar} | Kapasitas: {self.kapasitas}")
        print(f"   Harga Dasar/Malam: {format_rupiah(harga)} | Pajak(5%): {format_rupiah(pajak)}")
        print(f"   Menginap: {jumlah_malam} malam | Subtotal: {format_rupiah(subtotal)}")
        return subtotal


# =========================================================
# FITUR PEMESANAN: Proses transaksi dari daftar campuran kamar
# =========================================================
def proses_transaksi(daftar_pesanan):
    print("\n--- STRUK PEMESANAN ---")
    total_tagihan = 0
    for i, (kamar, jumlah_malam) in enumerate(daftar_pesanan, start=1):
        subtotal = kamar.tampilkan_detail(i, jumlah_malam)
        total_tagihan += subtotal
        print()
    print("----------------------------------------")
    print(f"TOTAL TAGIHAN: {format_rupiah(total_tagihan)}")
    print("----------------------------------------")


# =========================================================
# ALUR PROGRAM (USER STORY)
# =========================================================

# unit = KamarHotel("Test", "Test", 100000)  # ERROR! Abstract class tidak bisa diinstansiasi

print("--- SETUP DATA KAMAR ---")

# a) Admin membuat data kamar (1 Kamar Deluxe, 1 Kamar Standard)
kamar_deluxe = KamarDeluxe("Kamar Deluxe Sea View", 1_500_000, "Private Pool")
kamar_standard = KamarStandard("Kamar Standard Superior", 500_000, "2 Orang")

kamar_deluxe.tambah_stok(10)

# b) Admin mencoba mengisi stok kamar dengan angka negatif -> harus ditolak
kamar_standard.tambah_stok(-5)

# lanjut mengisi stok yang benar
kamar_standard.tambah_stok(20)

# c) Tamu memesan 2 malam Kamar Deluxe dan 1 malam Kamar Standard
daftar_pesanan = [
    (kamar_deluxe, 2),
    (kamar_standard, 1),
]

# d) Tampilkan detail kamar yang dipesan + total tagihan akhir
proses_transaksi(daftar_pesanan)
```
