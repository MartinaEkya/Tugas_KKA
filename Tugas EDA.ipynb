```python
import numpy as np
harga = np.array([5000, 7000, 3000, 12000, 4500])
print('Rata-rata harga:', harga.mean())
print('Harga tertinggi:', harga.max())
print('Harga setelah diskon 10%:', harga * 0.9)
```

Tugas Analisis 1: Bandingkan hasil harga * 0.9 di atas dengan jika harga berupa list Python
biasa (bukan array). Apa yang terjadi bila kamu mencoba [5000,7000,3000] * 0.9? Jelaskan mengapa berbeda

Kalau harga array, langsung ngalikan tiap elemennya sekaligus, hasilnya [ 4500.  6300.  2700. 10800.  4050.] Tapi kalau harga itu diganti list Python biasa dia malah error, soalnya operator * di list artinya "mengulang isi list", bukan mengalikan nilainya dan list cuma bisa diulang pakai bilangan bulat, bukan desimal seperti 0.9


Bedanya karena NumPy array memang dirancang buat operasi matematis ke semua elemen sekaligus (vectorized), sedangkan list Python itu struktur data umum yang nggak ngerti perkalian matematis kayak gitu harus di loop manual satu-satu kalau mau pakai list biasa

```python
import pandas as pd
data_kantin = {
'menu': ['Nasi Goreng', 'Es Teh', 'Mie Ayam', 'Es Teh', None],
'harga': [12000, 4000, 10000, 4000, 8000],
'terjual': [23, 40, None, 35, 18]
}
df = pd.DataFrame(data_kantin)
print(df)
```

Tugas Analisis 2: Amati tabel yang dihasilkan. Kolom mana yang terlihat memiliki data kosong(None)?
Menurutmu apa risikonya jika data ini langsung dianalisis tanpa dibersihkan?

Terdapat pada kolom menu dan kolom terjual

Jika data langsung dianalisis tanpa dibersihkan, hasil analisis dapat menjadi tidak akurat, terutama ketika menghitung jumlah penjualan atau pendapatan. Data kosong juga dapat menyebabkan kesalahan atau memengaruhi hasil perhitungan statistik

```python
df = pd.read_csv('data_kantin.csv')
print(df.head()) # 5 baris pertama
print(df.info()) # tipe data & jumlah non-null tiap kolom
print(df.describe()) # statistik ringkas kolom numerik
print(df.shape) # jumlah (baris, kolom)
```

Tugas analisis 3
Dari hasil df.info(), kolom mana yang jumlah non-null-nya lebih sedikit dari jumlah baris total? Apa artinya?

Kolom menu dan terjual memiliki jumlah non-null yang lebih sedikit. Artinya, di kedua kolom tersebut terdapat data yang kosong atau missing value

```python
print(df.isnull().sum()) # jumlah data kosong tiap kolom
df['terjual'] = df['terjual'].fillna(0) # isi kekosongan dengan 0
df = df.dropna(subset=['menu']) # hapus baris jika kolom menu kosong
```

Tugas Analisis 4
Soal: Mengapa terjual diisi dengan fillna, sedangkan baris dengan menu kosong dihapus?

Karena data terjual yang kosong masih bisa dianggap 0 penjualan, sehingga bisa diisi dengan 0. Sedangkan kalau menu kosong, kita tidak tahu data tersebut merupakan menu apa, jadi lebih masuk akal untuk menghapus barisnya

```python
print(df.duplicated().sum()) # jumlah baris duplikat
df = df.drop_duplicates()
df['harga'] = df['harga'].astype(int) # memastikan tipe data harga adalah integer
print(df.dtypes)
```

Tugas Analisis 5
Soal: Setelah drop_duplicates(), berapa jumlah baris dataset sekarang? Mengapa tipe data harus benar?

Setelah data duplikat dihapus, jumlah baris menjadi 102 baris. Tipe data harus benar supaya data bisa dihitung dan dianalisis dengan tepat. Misalnya, harga harus berupa angka agar bisa digunakan untuk menghitung pendapatan

```python
laris = df[df['terjual'] > 20] # filtering
urut = df.sort_values(by='terjual', ascending=False) # sorting
df['total_pendapatan'] = df['harga'] * df['terjual'] # kolom turunan
ringkasan = df.groupby('menu')['total_pendapatan'].sum() # agregasi
print(ringkasan)
```
